## Exercițiul 2

Generați și afișați grafic o sinusoidă de frecvență și fază aleasă de voi, dar ale cărei valori minime și maxime (atinse) să fie 3, respectiv 7.
     

```python

frecventa = 0.1  
faza = 0  
amplitudine = (7 - 3) / 2  
offset = (7 + 3) / 2  


timp = np.linspace(0, 50, 500)  
sinusoid = amplitudine * np.sin(2 * np.pi * frecventa * timp + faza) + offset


plt.figure(figsize=(10, 5))
plt.plot(timp, sinusoid, label=f'Sinusoidă (f = {frecventa} Hz)')
plt.xlabel('Timp (s)')
plt.ylabel('Amplitudine')
plt.title('Sinusoidă cu valori maxime de 7 și minime de 3')
plt.ylim(2, 8) 
plt.grid(True)
plt.legend()
plt.show()
```
