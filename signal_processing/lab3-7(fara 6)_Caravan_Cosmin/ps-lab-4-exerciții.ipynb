## Exercițiul 2

Frecvențele emise de un contrabas se încadrează între 40Hz și 200Hz.

Care este **frecvența minimă** cu care trebuie eșantionat semnalul provenit din înregistrarea instrumentului, astfel încât semnalul discretizat să conțină toate componentele de frecvență pe care instrumentul le poate produce?

```python

# Frecvența minimă de eșantionare necesară pentru a capta toate frecvențele produse de un contrabas
# (între 40 Hz și 200 Hz) este de 400 Hz. 
# Aceasta se bazează pe teorema Nyquist, care spune că frecvența de eșantionare trebuie să fie de cel
# puțin două ori mai mare decât frecvența maximă a semnalului. 
# Așadar, 2 x 200 Hz = 400 Hz.
```

## Exercițiul 3

Construiți un semnal sinusoidal de frecvență aleasă de voi, de amplitudine unitară și fază nulă. Realizați un grafic care să arăte că eșantionarea lui cu o **frecvență sub-Nyquist** (aleasă, de asemenea, de voi) generează fenomenul de aliere.

În acest scop, generați alte două semnale sinusoidale (cu frecvențe fundamentale _diferite_ față de cea a semnalului inițial) care eșantionate cu frecvența aleasă mai sus produc **aceleași eșantioane** ca semnalul inițial. Ar trebui să obțineți o situație similară cu cea din Figura 2 din PDF.

```python
frequency = 50 
sampling_frequency = 60 
duration = 0.1  
t = np.arange(0, duration, 1 / sampling_frequency)  

signal_original = np.sin(2 * np.pi * frequency * t)

frequency_alias_1 = sampling_frequency - frequency 
frequency_alias_2 = sampling_frequency + frequency

signal_alias_1 = np.sin(2 * np.pi * frequency_alias_1 * t)
signal_alias_2 = np.sin(2 * np.pi * frequency_alias_2 * t)

plt.figure(figsize=(12, 6))
plt.plot(t, signal_original, label=f'Semnal original {frequency} Hz', marker='o')
plt.plot(t, signal_alias_1, label=f'Semnal alias {frequency_alias_1} Hz', linestyle='--', marker='x')
plt.plot(t, signal_alias_2, label=f'Semnal alias {frequency_alias_2} Hz', linestyle=':', marker='s')
plt.xlabel("Timp (s)")
plt.ylabel("Amplitudine")
plt.title("Eșantionare sub-Nyquist și fenomenul de aliere")
plt.legend()
plt.grid(True)
plt.show()
```
