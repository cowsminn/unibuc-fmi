## 1. Importing Libraries

First, import the necessary libraries: pandas, numpy, matplotlib for visualization, and the relevant modules from scikit-learn.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

from sklearn.datasets import load_iris
from sklearn.cluster import KMeans

# For reproducibility
np.random.seed(42)
```

## 2. Loading the Iris Dataset

We will load the Iris dataset using scikit-learn. This dataset contains 150 samples and 4 features (sepal length, sepal width, petal length, and petal width). Although the data has species labels, we will use only the features to perform unsupervised clustering.

```python
# Load the Iris dataset
iris = load_iris()

# Create a DataFrame with the feature data
df_iris = pd.DataFrame(data=iris.data, columns=iris.feature_names)

# Optionally, add the target species for reference (not used in clustering)
df_iris['species'] = iris.target

df_iris.head()
```

## 3. Visualizing the Data

Before applying K-Means clustering, let's visualize the data with a scatter plot. Here we'll plot **sepal length** against **sepal width**. You can experiment with other feature combinations as well.

```python
# Scatter plot: Sepal Length vs Sepal Width
plt.figure(figsize=(8,6))
plt.scatter(df_iris['sepal length (cm)'], df_iris['sepal width (cm)'], c='blue', edgecolor='k', s=50)
plt.xlabel('Sepal Length (cm)')
plt.ylabel('Sepal Width (cm)')
plt.title('Iris Data: Sepal Length vs Sepal Width')
plt.show()
```

## 4. Applying K-Means Clustering

Now, we apply K-Means clustering to the dataset. Since the Iris dataset has 3 species, we set **k = 3**. K-Means will partition the data into 3 clusters based solely on the feature values.

```python
# Set the number of clusters
k = 3

# Initialize the K-Means model
kmeans = KMeans(n_clusters=k, random_state=42)

# Fit the model to the data (using only the feature values)
kmeans.fit(iris.data)

# Get the cluster labels
clusters = kmeans.labels_

# Add the cluster labels to our DataFrame
df_iris['cluster'] = clusters

# Display the first few rows with the cluster assignments
df_iris.head()
```

## 5. Visualizing the Clusters

Let's see how the K-Means algorithm has partitioned the data. In the scatter plot below, the color of each point corresponds to the cluster assigned by K-Means.

```python
# Scatter plot: Coloring points by cluster assignment
plt.figure(figsize=(8,6))
plt.scatter(df_iris['sepal length (cm)'], df_iris['sepal width (cm)'], c=df_iris['cluster'], cmap='viridis', edgecolor='k', s=50)
plt.xlabel('Sepal Length (cm)')
plt.ylabel('Sepal Width (cm)')
plt.title('K-Means Clustering of Iris Data')
plt.show()
```

## 6. Evaluating the Clustering

After clustering, K-Means provides some useful information such as the cluster centers and the **inertia** (the sum of squared distances between each sample and its nearest cluster center). Lower inertia generally indicates better clustering (though it tends to decrease with more clusters).

```python
# Output the cluster centers
print('Cluster Centers:')
print(kmeans.cluster_centers_)

# Output the inertia value
print('\nInertia:', kmeans.inertia_)
```
