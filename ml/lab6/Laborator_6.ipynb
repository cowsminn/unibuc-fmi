```python
M = np.zeros((10,784,4))

for i in range(10):
    
    indexes_of_class_i = np.ravel(np.where(train_labels == i))
    
    for j in range(784):
        valoriPixeli = train_images[indexes_of_class_i, j]
        histogram, _ = np.histogram(valoriPixeli, limiteInterval)
        
        M[i,j,:] = histogram / np.sum(histogram)
        
M = M + 0.0000001



```

```python
from math import *

def classifierNaiveBayes(img, M, limiteInterval, p_c):
    prob_img = np.zeros(10)
    for i in range(10):
        prob_img[i] = log(p_C[i])
        for j in range(784):
            intervalul = np.digitize(img[j], limiteInterval)
            prob_img[i] += log(M[i, j, intervalul-1])
    predicted_class = prob_img.argmax()
    return predicted_class
```

```python
image = test_images[1, :]
plt.imshow(image.reshape(28,28), cmap = "gray")
plt.show()
classifierNaiveBayes(image, M, limiteInterval, p_C)
```

```python
# Calculate class prior probabilities
p_C = np.zeros(10)
for i in range(10):
    p_C[i] = np.sum(train_labels == i) / len(train_labels)

# Evaluate on test data
correct_predictions = 0
for idx, test_img in enumerate(test_images):
    predicted = classifierNaiveBayes(test_img, M, limiteInterval, p_C)
    if predicted == test_labels[idx]:
        correct_predictions += 1

accuracy = correct_predictions / len(test_labels)
print(f"Accuracy: {accuracy:.4f}")
```

```python
import numpy as np
from sklearn.metrics import confusion_matrix
import matplotlib.pyplot as plt
import seaborn as sns

# Inițializăm liste pentru a stoca etichetele adevărate și cele prezise
y_true = []
y_pred = []

# Calculăm predicțiile pentru toate imaginile de test
for idx, test_img in enumerate(test_images):
    predicted = classifierNaiveBayes(test_img, M, limiteInterval, p_C)
    y_true.append(test_labels[idx])
    y_pred.append(predicted)

# Calculăm matricea de confuzie
conf_matrix = confusion_matrix(y_true, y_pred)

# Afișăm matricea de confuzie ca o matrice numerică
print("Matricea de confuzie:")
print(conf_matrix)

# Vizualizăm matricea de confuzie
plt.figure(figsize=(10, 8))
sns.heatmap(conf_matrix, annot=True, fmt='d', cmap='Blues',
            xticklabels=range(10), yticklabels=range(10))
plt.xlabel('Clasa prezisă')
plt.ylabel('Clasa adevărată')
plt.title('Matricea de confuzie (10x10)')
plt.show()
```
